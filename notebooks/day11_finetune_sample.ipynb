```python
# Cell 2:对质谱数据进行向量化
def binned_spectrum(mz_list,inten_list,mz_max=1000,bin_size=1.0):
    #初始化向量
    size=int(mz_max/bin_size);
    vec=np.zeros(size,dtype=np.float32);

    for mz,inten in zip(mz_list,inten_list):
        if 0<=mz<mz_max:#注意0可以取到但是mz_max取不到
            idx=int(mz/bin_size);#错误：这里是bin_size不是size
            vec[idx]=max(vec[idx],float(inten));#位置冲突，存储最大的强度

    #再做一次归一化处理
    if vec.max()>0:
        vec=vec/vec.max();

    return vec;

#测试一下函数功能有无问题;
test_mz = [194.12, 212.13, 730.42, 195.12];
test_inten = [1.0, 0.58, 0.56, 0.54];
test_vec = binned_spectrum(test_mz, test_inten);
print("测试谱向量形状:", test_vec.shape);
print("非零位置:", np.nonzero(test_vec)[0]);
print("非零强度:", test_vec[np.nonzero(test_vec)[0]]);
```

```python
#Cell 3:创建谱编码器
def build_spec_encoder():
    model=nn.Sequential(
        nn.Linear(1000,512),
        nn.ReLU(),
        nn.Linear(512,256),
        nn.ReLU(),
        nn.Linear(256,2048)
    );

    return model;
```

```python
spec_encoder = build_spec_encoder();

print("谱编码器模型结构:");
print(spec_encoder);

#### 测试输入输出形状
test_input = torch.tensor(test_vec, dtype=torch.float32);
test_output = spec_encoder(test_input);

print("输入谱图形状:", test_input.shape);
print("输出条件形状:", test_output.shape);
```

```python
#Cell 5:加载 CANOPUS 小数据集
#### 一、设置数据路径
labels_path = os.path.join(data_dir, "labels.tsv");
spec_dir = os.path.join(data_dir, "subformulae", "subformulae_default");

print("标签文件路径:", labels_path);
print("谱图 json 文件夹:", spec_dir);
print("标签文件是否存在:", os.path.exists(labels_path));
print("谱图文件夹是否存在:", os.path.exists(spec_dir));


#### 二、读取数据并整理成训练样本
max_samples = 500;

all_data = [];
missing_json = 0;
bad_mol = 0;

with open(labels_path, "r", encoding="utf-8") as f:
    reader = csv.DictReader(f, delimiter="\t");

    for row in reader:
        # 有的数据表字段叫 spec，有的叫 specname，这里做一个兼容
        if "spec" in row:
            spec = row["spec"];
        else:
            spec = row["specname"];

        smiles = row["smiles"];
        json_path = os.path.join(spec_dir, spec + ".json");

        # 如果找不到对应谱图文件，就跳过
        if not os.path.exists(json_path):
            missing_json += 1;
            continue;

        # 读取 json 里的 mz 和 intensity
        with open(json_path, "r", encoding="utf-8") as jf:
            obj = json.load(jf);

        mz_list = obj["output_tbl"]["mz"];
        inten_list = obj["output_tbl"]["ms2_inten"];
        spec_vec = binned_spectrum(mz_list, inten_list);

        # 把 SMILES 转成 RDKit 分子
        mol = Chem.MolFromSmiles(smiles);
        if mol is None:
            bad_mol += 1;
            continue;

        # 把分子转成邻接矩阵、原子类型、分子指纹
        A0, atom_symbols = build_adj(mol);
        atom_types = get_atom_one_hot(atom_symbols);
        fp = get_fingerprint(mol);

        all_data.append({
            "spec": spec,
            "smiles": smiles,
            "spec_vec": spec_vec,
            "A0": A0,
            "A0_labels": np.argmax(A0, axis=-1),
            "atom_types": atom_types,
            "fp": fp,
            "n": A0.shape[0]
        });

        if len(all_data) >= max_samples:
            break;

#### 三、查看加载结果
print("加载完成:", len(all_data), "条样本");
print("谱文件缺失:", missing_json);
print("分子解析失败:", bad_mol);

first = all_data[0];

print("第一条 spec:", first["spec"]);
print("第一条 smiles:", first["smiles"]);
print("第一条谱向量形状:", first["spec_vec"].shape);
print("第一条 A0 形状:", first["A0"].shape);
print("第一条 atom_types 形状:", first["atom_types"].shape);
print("第一条指纹形状:", first["fp"].shape);
```
